### Red en T:


$$Y_1 = \frac{1}{sL}, \quad Y_2 = \frac{1}{sL}, \quad Y_3 = sC$$



#### 1. Supresion

$$\begin{bmatrix} I' \\ \hline 0 \end{bmatrix} = \begin{bmatrix} [Y_{11}] & [Y_{12}] \\ \hline [Y_{21}] & [Y_{22}] \end{bmatrix} \begin{bmatrix} V' \\ \hline V_d \end{bmatrix}$$

Donde:
* $V' = \begin{bmatrix} V_1 \\ V_2 \\ V_3 \end{bmatrix}, \quad I' = \begin{bmatrix} I_1 \\ I_2 \\ I_3 \end{bmatrix}$
* $V_d = [V_4], \quad I_d = [0]$

Identificando las submatrices por inspección:

$$[Y_{11}] = \begin{bmatrix} Y_1 & 0 & 0 \\ 0 & Y_2 & 0 \\ 0 & 0 & Y_3 \end{bmatrix} = \begin{bmatrix} \frac{1}{sL} & 0 & 0 \\ 0 & \frac{1}{sL} & 0 \\ 0 & 0 & sC \end{bmatrix}$$

$$[Y_{12}] = \begin{bmatrix} -Y_1 \\ -Y_2 \\ -Y_3 \end{bmatrix} = \begin{bmatrix} -\frac{1}{sL} \\ -\frac{1}{sL} \\ -sC \end{bmatrix}$$

$$[Y_{21}] = \begin{bmatrix} -Y_1 & -Y_2 & -Y_3 \end{bmatrix} = \begin{bmatrix} -\frac{1}{sL} & -\frac{1}{sL} & -sC \end{bmatrix}$$

$$[Y_{22}] = [Y_1 + Y_2 + Y_3] = \left[ \frac{1}{sL} + \frac{1}{sL} + sC \right] = \left[ \frac{2}{sL} + sC \right] = \left[ \frac{2 + s^2LC}{sL} \right]$$

---



$$Y_{22}^{-1} = \frac{1}{\frac{2 + s^2LC}{sL}} = \frac{sL}{s^2LC + 2}$$

---

#### $Y_{12} Y_{22}^{-1} Y_{21}$

$$Y_{12} Y_{22}^{-1} Y_{21} = \frac{sL}{s^2LC + 2} \begin{bmatrix} -\frac{1}{sL} \\ -\frac{1}{sL} \\ -sC \end{bmatrix} \begin{bmatrix} -\frac{1}{sL} & -\frac{1}{sL} & -sC \end{bmatrix}$$



$$Y_{12} Y_{22}^{-1} Y_{21} = \frac{sL}{s^2LC + 2} \begin{bmatrix} \frac{1}{s^2L^2} & \frac{1}{s^2L^2} & \frac{C}{L} \\ \frac{1}{s^2L^2} & \frac{1}{s^2L^2} & \frac{C}{L} \\ \frac{C}{L} & \frac{C}{L} & s^2C^2 \end{bmatrix} = \frac{1}{s^2LC + 2} \begin{bmatrix} \frac{1}{sL} & \frac{1}{sL} & sC \\ \frac{1}{sL} & \frac{1}{sL} & sC \\ sC & sC & s^3LC^2 \end{bmatrix}$$

---

 $Y_{\text{MAI}} = Y_{11} - Y_{12} Y_{22}^{-1} Y_{21}$

$$Y_{\text{MAI}} = \begin{bmatrix} \frac{1}{sL} & 0 & 0 \\ 0 & \frac{1}{sL} & 0 \\ 0 & 0 & sC \end{bmatrix} - \frac{1}{s^2LC + 2} \begin{bmatrix} \frac{1}{sL} & \frac{1}{sL} & sC \\ \frac{1}{sL} & \frac{1}{sL} & sC \\ sC & sC & s^3LC^2 \end{bmatrix}$$


  $$y_{11} = y_{22} = \frac{1}{sL} - \frac{\frac{1}{sL}}{s^2LC + 2} = \frac{1}{sL} \left( 1 - \frac{1}{s^2LC + 2} \right) = \frac{1}{sL} \left( \frac{s^2LC + 1}{s^2LC + 2} \right)$$

  $$y_{33} = sC - \frac{s^3LC^2}{s^2LC + 2} = sC \left( 1 - \frac{s^2LC}{s^2LC + 2} \right) = sC \left( \frac{2}{s^2LC + 2} \right) = \frac{2sC}{s^2LC + 2}$$


  $$y_{12} = y_{21} = -\frac{1}{sL(s^2LC + 2)}$$
  $$y_{13} = y_{31} = y_{23} = y_{32} = -\frac{sC}{s^2LC + 2}$$

---

#### Matriz Admitancia Indefinida

$$Y_{\text{MAI}} = \frac{1}{s^2LC + 2} \begin{bmatrix}
\frac{s^2LC + 1}{sL} & -\frac{1}{sL} & -sC \\
-\frac{1}{sL} & \frac{s^2LC + 1}{sL} & -sC \\
-sC & -sC & 2sC
\end{bmatrix}$$

---

# 2. Funciones de Transferencia



$$T = \frac{V_{ij}}{V_{mn}} = \operatorname{(signo)}\cdot \frac{Y_{mn}^{ij}}{Y_{mn}^{mn}}$$



---

### $T_1 = \dfrac{V_{23}}{V_{13}}$

  * Denominador:
    $$Y_{13}^{13} = y_{22} = \frac{1 + s^2LC}{sL(s^2LC + 2)}$$

* **Cociente:**
  $$\frac{V_{23}}{V_{13}} = \frac{Y_{13}^{23}}{Y_{13}^{13}} = \frac{\frac{1}{sL(s^2LC + 2)}}{\frac{1 + s^2LC}{sL(s^2LC + 2)}} = \mathbf{\frac{1}{s^2LC + 1}}$$


---

###$T_2 = \dfrac{V_{21}}{V_{31}}$


  * Numerador:
    $$Y_{31}^{21} = y_{23} = \frac{s^2LC}{sL(s^2LC + 2)}$$
  * Denominador:
    $$Y_{31}^{31} = y_{22} = \frac{1 + s^2LC}{sL(s^2LC + 2)}$$

* **Cociente:**
  $$\frac{V_{21}}{V_{31}} = \frac{Y_{31}^{21}}{Y_{31}^{31}} = \frac{\frac{s^2LC}{sL(s^2LC + 2)}}{\frac{1 + s^2LC}{sL(s^2LC + 2)}} = \mathbf{\frac{s^2LC}{s^2LC + 1}}$$


---

### $T_3 = \dfrac{V_{12}}{V_{32}}$


* Por simetría estructural de la red respecto a los terminales 1 y 2:
  $$\frac{V_{12}}{V_{32}} = \frac{Y_{32}^{12}}{Y_{32}^{32}} = \mathbf{\frac{s^2LC}{s^2LC + 1}}$$

---

# 3. Análisis de Complementariedad en Módulo


$$T_a(s) + T_b(s) = \frac{1}{s^2LC + 1} + \frac{s^2LC}{s^2LC + 1} = \frac{s^2LC + 1}{s^2LC + 1} = 1$$

---

### 2. Evaluación en régimen permanente armónico ($s = j\omega$):

$$|T_a(j\omega) + T_b(j\omega)| = |1| = 1 \quad \forall \, \omega \in \mathbb{R}$$



```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import signal


L = 1e-3  # Inductancia (1 mH)
C = 1e-6  # Capacitancia (1 uF)

w0 = 1.0 / np.sqrt(L * C)
f0 = w0 / (2 * np.pi)

print(f"Frecuencia de resonancia w0: {w0:.2f} rad/s (~{f0:.2f} Hz)")


den = [L * C, 0, 1]

# T1(s) = 1 / (LC*s^2 + 1)
num_T1 = [1]
sys_T1 = signal.TransferFunction(num_T1, den)

# T2(s) = T3(s) = (LC*s^2) / (LC*s^2 + 1)
num_T2 = [L * C, 0, 0]
sys_T2 = signal.TransferFunction(num_T2, den)


w = np.logspace(np.log10(w0) - 2, np.log10(w0) + 2, 2000)

w_T1, mag_T1, phase_T1 = signal.bode(sys_T1, w=w)
w_T2, mag_T2, phase_T2 = signal.bode(sys_T2, w=w)

fig, (ax_mag, ax_phase) = plt.subplots(
    2, 1, figsize=(10, 8), sharex=True, dpi=100
)

# --- Gráfico de Magnitud (dB) ---
ax_mag.semilogx(w, mag_T1, label=r"$T_1(s) = \frac{1}{s^2LC + 1}$ (Pasa-bajos)", color="tab:blue", lw=2)
ax_mag.semilogx(w, mag_T2, label=r"$T_2(s) = T_3(s) = \frac{s^2LC}{s^2LC + 1}$ (Pasa-altos)", color="tab:orange", lw=2, linestyle="--")
ax_mag.axvline(w0, color="gray", linestyle=":", label=r"$\omega_0 = 1/\sqrt{LC}$")
ax_mag.set_ylabel("Módulo [dB]", fontsize=12)
ax_mag.set_title("Respuesta en Frecuencia (Diagrama de Bode)", fontsize=14, fontweight="bold")
ax_mag.grid(True, which="both", ls="-", alpha=0.5)
ax_mag.legend(loc="lower left", fontsize=10)
ax_mag.set_ylim(-60, 40)

# --- Gráfico de Fase (Grados) ---
ax_phase.semilogx(w, phase_T1, label=r"$T_1(s)$", color="tab:blue", lw=2)
ax_phase.semilogx(w, phase_T2, label=r"$T_2(s) = T_3(s)$", color="tab:orange", lw=2, linestyle="--")
ax_phase.axvline(w0, color="gray", linestyle=":", label=r"$\omega_0$")
ax_phase.set_xlabel(r"Frecuencia angular $\omega$ [rad/s]", fontsize=12)
ax_phase.set_ylabel("Fase [grados]", fontsize=12)
ax_phase.set_yticks([-180, -90, 0, 90, 180])
ax_phase.grid(True, which="both", ls="-", alpha=0.5)
ax_phase.legend(loc="center left", fontsize=10)

plt.tight_layout()
plt.show()
```

Las relaciones con el circuito equivalente son:

$$C_G = C_{iss} - C_{rss} = 24\text{ pF} - 7\text{ pF} = 17\text{ pF}$$
$$C_D = C_{oss} - C_{rss} = 17\text{ pF} - 7\text{ pF} = 10\text{ pF}$$

# Obtención de la Matriz de Admitancia Indefinida (MAI) a partir de la Matriz de Admitancia Definida (MAD)

---

### 1. Planteo de Ecuaciones de Nodo para la Matriz Definida (MAD)

Tomando el terminal **Source (S)** como nodo de referencia ($V_S = 0$), definimos las tensiones de los puertos respecto de masa:
* Tensión de Gate: $V_1 = V_G$
* Tensión de Drain: $V_2 = V_D$

Planteamos las corrientes entrantes a cada nodo ($I_1$ en Gate e $I_2$ en Drain):

* **En el nodo Gate (1):**
$$I_1 = I_{C_G} + I_{C_{GD}}$$
$$I_1 = s C_G V_1 + s C_{GD} (V_1 - V_2)$$
$$I_1 = s(C_G + C_{GD}) V_1 - s C_{GD} V_2$$

* **En el nodo Drain (2):**
$$I_2 = g_m V_1 + I_{C_D} + I_{C_{GD}}'$$
$$I_2 = g_m V_1 + s C_D V_2 + s C_{GD} (V_2 - V_1)$$
$$I_2 = (g_m - s C_{GD}) V_1 + s(C_D + C_{GD}) V_2$$

En forma matricial para el cuadripolo ($2 \times 2$):

$$\begin{bmatrix} I_1 \\ I_2 \end{bmatrix} = \begin{bmatrix} y_{11} & y_{12} \\ y_{21} & y_{22} \end{bmatrix} \begin{bmatrix} V_1 \\ V_2 \end{bmatrix}$$

Donde la **Matriz de Admitancia Definida** es:

$$[Y_{\text{MAD}}] = \begin{bmatrix}
s(C_G + C_{GD}) & -s C_{GD} \\
g_m - s C_{GD} & s(C_D + C_{GD})
\end{bmatrix}$$


---

### 2. Expansión a la Matriz de Admitancia Indefinida (MAI)

La MAI describe la red completa de 3 terminales: **Gate (1)**, **Drain (2)** y **Source (3)**:

$$[Y_{\text{MAI}}] = \begin{bmatrix}
y_{GG} & y_{GD} & y_{GS} \\
y_{DG} & y_{DD} & y_{DS} \\
y_{SG} & y_{SD} & y_{SS}
\end{bmatrix}$$

Por definición, la submatriz correspondiente a los nodos independientes respecto de la referencia coincide con la MAD:

$$y_{GG} = y_{11} = s(C_G + C_{GD})$$
$$y_{GD} = y_{12} = -s C_{GD}$$
$$y_{DG} = y_{21} = g_m - s C_{GD}$$
$$y_{DD} = y_{22} = s(C_D + C_{GD})$$

---



### 4. Matriz de Admitancia Indefinida Final

$$[Y_{\text{MAI}}] = \begin{bmatrix}
s(C_G + C_{GD}) & -s C_{GD} & -s C_G \\
g_m - s C_{GD} & s(C_D + C_{GD}) & -(g_m + s C_D) \\
-(g_m + s C_G) & -s C_D & g_m + s(C_G + C_D)
\end{bmatrix}$$

En función directa de los parámetros del datasheet ($C_{iss}, C_{rss}, C_{oss}$):

$$[Y_{\text{MAI}}] = \begin{bmatrix}
s C_{iss} & -s C_{rss} & -s(C_{iss} - C_{rss}) \\
g_m - s C_{rss} & s C_{oss} & -[g_m + s(C_{oss} - C_{rss})] \\
-[g_m + s(C_{iss} - C_{rss})] & -s(C_{oss} - C_{rss}) & g_m + s(C_{iss} + C_{oss} - 2C_{rss})
\end{bmatrix}$$

Reemplazando los valores numéricos ($C$ en $\text{pF}$):

$$[Y_{\text{MAI}}] = \begin{bmatrix}
24 \cdot 10^{-12} s & -7 \cdot 10^{-12} s & -17 \cdot 10^{-12} s \\
g_m - 7 \cdot 10^{-12} s & 17 \cdot 10^{-12} s & -(g_m + 10 \cdot 10^{-12} s) \\
-(g_m + 17 \cdot 10^{-12} s) & -10 \cdot 10^{-12} s & g_m + 27 \cdot 10^{-12} s
\end{bmatrix}$$

---

#Parámetros $Y$ para Configuraciones Source Común y Drain Común

###Source Común (SC)
Tomando el terminal **Source (3)** como referencia, se eliminan la fila 3 y la columna 3 de la MAI. Los puertos son Entrada: Gate (1), Salida: Drain (2):

$$[Y_{SC}] = \begin{bmatrix}
y_{11} & y_{12} \\
y_{21} & y_{22}
\end{bmatrix} = \begin{bmatrix}
24 \cdot 10^{-12} s & -7 \cdot 10^{-12} s \\
g_m - 7 \cdot 10^{-12} s & 17 \cdot 10^{-12} s
\end{bmatrix}$$

###Drain Común (DC) / Seguidor por Fuente
Tomando el terminal **Drain (2)** como referencia, se eliminan la fila 2 y la columna 2 de la MAI. Los puertos son Entrada: Gate (1), Salida: Source (3):

$$[Y_{DC}] = \begin{bmatrix}
y_{GG} & y_{GS} \\
y_{SG} & y_{SS}
\end{bmatrix} = \begin{bmatrix}
24 \cdot 10^{-12} s & -17 \cdot 10^{-12} s \\
-(g_m + 17 \cdot 10^{-12} s) & g_m + 27 \cdot 10^{-12} s
\end{bmatrix}$$

---

#Matriz ($ABCD$) en Source Común

A partir de los parámetros de admitancia en configuración Source Común $[Y_{SC}]$, las relaciones de conversión a la matriz de transmisión $T = \begin{bmatrix} A & B \\ C & D \end{bmatrix}$ (donde $V_1 = A V_2 - B I_2$ e $I_1 = C V_2 - D I_2$) son:

$$A = -\frac{y_{22}}{y_{21}} = -\frac{17 \cdot 10^{-12} s}{g_m - 7 \cdot 10^{-12} s}$$

$$B = -\frac{1}{y_{21}} = -\frac{1}{g_m - 7 \cdot 10^{-12} s}$$

$$C = -\frac{\det(Y_{SC})}{y_{21}} = -\frac{y_{11}y_{22} - y_{12}y_{21}}{y_{21}}$$

Calculando el determinante:
$$\det(Y_{SC}) = (24 \cdot 10^{-12} s)(17 \cdot 10^{-12} s) - (-7 \cdot 10^{-12} s)(g_m - 7 \cdot 10^{-12} s)$$
$$\det(Y_{SC}) = 408 \cdot 10^{-24} s^2 + 7 \cdot 10^{-12} s \cdot g_m - 49 \cdot 10^{-24} s^2$$
$$\det(Y_{SC}) = 7 \cdot 10^{-12} s \cdot g_m + 359 \cdot 10^{-24} s^2$$

Por lo tanto:
$$C = -\frac{7 \cdot 10^{-12} s \cdot g_m + 359 \cdot 10^{-24} s^2}{g_m - 7 \cdot 10^{-12} s}$$

$$D = -\frac{y_{11}}{y_{21}} = -\frac{24 \cdot 10^{-12} s}{g_m - 7 \cdot 10^{-12} s}$$

### Matriz ABCD Final:

$$[T] = \begin{bmatrix} A & B \\ C & D \end{bmatrix} = -\frac{1}{g_m - 7 \cdot 10^{-12} s} \begin{bmatrix}
17 \cdot 10^{-12} s & 1 \\
7 \cdot 10^{-12} s \cdot g_m + 359 \cdot 10^{-24} s^2 & 24 \cdot 10^{-12} s
\end{bmatrix}$$
